```python
import numpy as np 
import pandas as pd  
```

```python
df = pd.read_csv('startup_funding.csv')
df.head()
```

```python
sorted(df['Startup Name'].unique().tolist())
```

```python
df.head()
```

```python
df['Investors Name'].unique().tolist()
```

```python
df['Investors Name'] = df['Investors Name'].fillna('Undisclosed')
```

```python
sorted(df['Investors Name'].unique().tolist())
```

```python
df.info()
```

```python
df.drop('Remarks', axis=1, inplace=True)
```

```python
df.set_index('Sr No', inplace=True)
```

```python
df.head()
```

```python
df['Amount in USD'] = df['Amount in USD'].fillna('0')
```

```python
df.info()
```

```python
df['Amount in USD'] = df['Amount in USD'].str.replace(',', '')
```

```python
df['Amount in USD'] = df['Amount in USD'].str.replace('Undisclosed', '0')
df['Amount in USD'] = df['Amount in USD'].str.replace('undisclosed', '0')
```

```python
df['Amount in USD'] = df['Amount in USD'].str.replace('unknown', '0')
```

```python
df = df[df['Amount in USD'].str.isdigit()]
```

```python
df['Amount in USD'] = df['Amount in USD'].astype('float')
```

```python
df.info()
```

```python
def to_inr(dollar):
    inr = dollar * 80.5
    return inr / 10000000
```

```python
df['Amount in USD'] = df['Amount in USD'].apply(to_inr)
```

```python
df.rename(columns={'Amount in USD': 'Amount in INR (Cr)'}, inplace=True)
```

```python
df.info()
```

```python
df['Date dd/mm/yyyy'] = df['Date dd/mm/yyyy'].str.replace("05/072018", "05/07/2018")
```

```python
df['Date dd/mm/yyyy'] = df['Date dd/mm/yyyy'].str.replace("01/07/015", "01/07/2015")
```

```python
df['Date dd/mm/yyyy'] = df['Date dd/mm/yyyy'].str.replace("12/05.2015", "12/05/2015")
```

```python
df['Date dd/mm/yyyy'] = df['Date dd/mm/yyyy'].str.replace("13/04.2015", "13/04/2015")
```

```python
df['Date dd/mm/yyyy'] = df['Date dd/mm/yyyy'].str.replace("15/01.2015", "15/01/2015")
```

```python
df['Date dd/mm/yyyy'] = df['Date dd/mm/yyyy'].str.replace("22/01//2015", "22/01/2015")
```

```python
df['Date dd/mm/yyyy'] = pd.to_datetime(df['Date dd/mm/yyyy'], format='%d/%m/%Y')
```

```python
df.info()
```

```python
df.rename(columns={'Date dd/mm/yyyy': 'Date'}, inplace=True)
```

```python
df.head()
```

```python
df['Date'].dt.month_name().unique()
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['Date', 'Industry Vertical', 'City  Location', 'InvestmentnType'], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.info()
```

```python
df.to_csv('cleaned_startup_funding.csv', index=False)
```
