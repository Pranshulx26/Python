### `Q-4:` Perform `groupby` based on the `"Team"` column and then perform Z Normalization on top of the below columns of each group:
1. Passes
2. Passes Completed
3. Attempted Line Breaks
4. Completed Line Breaks

You have to make a python function named `z_normalization` which takes two arguments:

1. *group:* Every group that you have created
2. *cols_to_perform:* This parameter takes a list of columns on which you have to perform the Z-Normalization.

For this problem, you have to use th `apply()` method.

$\Large Z \ - \ Normalization = \frac{X_i - \mu}{std} $

After that find out the below values for each group:
- minimum "Passess"
- maximum "Passess"
- minimum "Yellow Cards"
- maximum "Yellow cards"
- average "Yellow Cards"
- maximum "Attempted Line Breaks"
- minimum "Attempted Line Breaks"
- standardard deviation of "Attempted Line Breaks"
- average Possession

````python
def z_normalization(group, cols_to_perform):
    for col in cols_to_perform:
        mean = group[col].mean()
        std = group[col].std()

        if std != 0:
            group[col] = (group[col] - mean) / std
        else:
            group[col] = 0.0
    return group 


cols = ['Passes', 'Passes Completed', 'Attempted Line Breaks', 'Completed Line Breaks']
df = fifa_df.groupby('Team').apply(z_normalization, cols_to_perform=cols)
````

````python
stats = df.groupby('Team').agg(
    min_passes=('Passes', 'min'),
    max_passes=('Passes', 'max'),
    min_yellow_cards=('Yellow Cards', 'min'),
    max_yellow_cards=('Yellow Cards', 'max'),
    avg_yellow_cards=('Yellow Cards', 'mean'),
    max_attempted_line_breaks=('Attempted Line Breaks', 'max'),
    min_attempted_line_breaks=('Attempted Line Breaks', 'min'),
    std_attempted_line_breaks=('Attempted Line Breaks', 'std'),
    avg_possession=('Possession (%)', 'mean')
)
````

###`Q-5:` Find batsman in below category-
* Highest score while chasing
* Best Strike rate while chasing and have faced 100+ balls


> Chasing mean team batting in second inning

````python
# code here
chasing_df = ipl[ipl['innings'] == 2]
chasing_df.head()
````

````python
scores_per_match = chasing_df.groupby(['ID', 'batter'])['batsman_run'].sum().reset_index()
scores_per_match
````

````python
highest_chase_idx = scores_per_match['batsman_run'].idxmax()
highest_scorer = scores_per_match.loc[highest_chase_idx]
````

````python
print(f"Highest Score while Chasing: {highest_scorer['batsman_run']} runs by {highest_scorer['batter']}")
````

````python
# Group by batter to get career chasing totals
batter_stats = chasing_df.groupby('batter').agg(
    total_runs=('batsman_run', 'sum'),
    balls_faced=('ballnumber', 'count')
)

# Filter for batters who have faced 100+ balls while chasing
qualified_batters = batter_stats[batter_stats['balls_faced'] >= 100].copy()

# Calculate Strike Rate: (Runs / Balls) * 100
qualified_batters['strike_rate'] = (qualified_batters['total_runs'] / qualified_batters['balls_faced']) * 100

# Find the batter with the highest strike rate
best_sr_batter = qualified_batters['strike_rate'].idxmax()
best_sr_value = qualified_batters['strike_rate'].max()

print(f"Best Strike Rate while Chasing (100+ balls): {best_sr_value:.2f} by {best_sr_batter}")
````

###`Q-8:` Make a dataframe for all batting pairs played together.
```
Batsman1 Batsman2 Runs Avg StrikeRate
```

> Just to ease this question you can count wide-balls for strike rate.

````python


# 1. Standardize the pair names alphabetically
def get_pair(row):
    return tuple(sorted([row['batter'], row['non-striker']]))

ipl['pair'] = ipl.apply(get_pair, axis=1)

# 2. Calculate Runs and Balls Faced (including wide-balls as per instructions)
# We count every row in the group as a ball faced
pair_stats = ipl.groupby('pair').agg(
    Runs=('total_run', 'sum'),
    Balls=('pair', 'count')
).reset_index()

# 3. Calculate Strike Rate: (Runs / Balls) * 100
pair_stats['StrikeRate'] = (pair_stats['Runs'] / pair_stats['Balls']) * 100

# 4. Calculate Average: Total Runs / Number of Innings played together
# We find unique (Match ID, Pair) combinations to count the innings
pair_innings = ipl.groupby(['ID', 'pair']).size().reset_index().groupby('pair').size().reset_index(name='Innings')

# 5. Merge and Finalise
final_df = pd.merge(pair_stats, pair_innings, on='pair')
final_df['Avg'] = final_df['Runs'] / final_df['Innings']

# Split the pair tuple into two separate columns
final_df[['Batsman1', 'Batsman2']] = pd.DataFrame(final_df['pair'].tolist(), index=final_df.index)

# Select and reorder columns
batting_pairs_df = final_df[['Batsman1', 'Batsman2', 'Runs', 'Avg', 'StrikeRate']]

print(batting_pairs_df.sort_values(by='Runs', ascending=False))
````
