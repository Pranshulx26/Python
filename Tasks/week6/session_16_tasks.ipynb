
````python
# code here
items = pd.read_csv('items.csv')
items.head()
````

````python
# 1. 
items_series = items.set_index('item_name')
items_series
````

````python
# 2. show number of nan values
items_series.isnull().sum()
````

````python
items_series.dropna(inplace=True)
````

````python
s = items_series['item_price'].str.split('$').apply(lambda x: x[1]).astype(float)*90
mean = s.mean()
mean
````
