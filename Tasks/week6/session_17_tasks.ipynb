###`Q-8:` Team1 vs Team2 Dashbord
Create a function which will take two string(name of two teams) as input. Show win Loss record between them and player getting most player of the match award in matches between these two teams.
```
team1_vs_team2('Kolkata Knight Riders','Chennai Super Kings')
```

````python
# code here
def record(t1, t2):
    if t1 not in ipl['Team1'].unique() or t2 not in ipl['Team1'].unique():
        return 'Team not exist'
    data = ipl[((ipl['Team1'] == t1) & (ipl['Team2'] == t2)) | ((ipl['Team1'] == t2) & (ipl['Team2'] == t1))]
    win_loss = data['WinningTeam'].value_counts()
    potm = data['Player_of_Match'].value_counts().head(1)

    return win_loss, potm

record('Rajasthan Royals', 'Mumbai Indians')
````
