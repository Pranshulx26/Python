###`Q-1`: You are given a function definition. There might be several issues on execution of this function. You are asked to do exception handling for diffrent errors that this function goes in to `without altering this function`. And print error text.



Function parameters `l -> list, s -> could be anything`

```
def function(l: list, s, **args):
    last_element = l[-1]
    
    l[int(s)]=10
    any_element = l[int(s)+10]
    l[s]=10
    
    res = sum(l)
    
    p = args['p']
    # print(p)
    return res/last_element * p + any_element

```
Check for different function calls:-

```
function([1,2,1], 12)
function([1,2,1]*9, '1-2')
function([1,'2',1]*9, 12)
function([1,'2',1]*9, 12)
function([1,2,0]*9, 12  )
function([1,2,1]*9, 12, p=None)
function([1,2,0]*9, 12, p=10)
```

````python
# Write code here
def function(l: list, s, **args):
    last_element = l[-1]

    l[int(s)]=10
    any_element = l[int(s)+10]
    l[s]=10

    res = sum(l)

    p = args['p']
    # print(p)
    return res/last_element * p + any_element

def safe_call(l, s, **args):
    try:
        result = function(l, s, **args)
        print("Result:", result)

    except IndexError as e:
        print("IndexError:", e)

    except ValueError as e:
        print("ValueError:", e)

    except TypeError as e:
        print("TypeError:", e)

    except KeyError as e:
        print("KeyError:", e)

    except ZeroDivisionError as e:
        print("ZeroDivisionError:", e)

    except Exception as e:
        print("Unexpected Error:", e)

safe_call([1,2,1], 12)
safe_call([1,2,1]*9, '1-2')
safe_call([1,'2',1]*9, 12)
safe_call([1,'2',1]*9, 12)
safe_call([1,2,0]*9, 12  )
safe_call([1,2,1]*9, 12, p=None)
safe_call([1,2,0]*9, 12, p=10)
````

l = [{0:2},2,3,4,'5', {5:10}]
# For calculating sum of above list

````python
# Write code here

l = [{0:2},2,3,4,'5', {5:10}]
# For calculating sum of above list
s=0
for i in range(len(l)):
  try:
    if isinstance(l[i], dict):
      s += list(l[i].values())[0]
    elif isinstance(l[i], int):
      s += l[i]
    elif isinstance(l[i], str) and l[i].isdigit():
      s += int(l[i])

  except Exception as e:
    print(e)


print(s)
````

### `Q-3:`: File Handling with Exception handling

Write a program that opens a text file and write data to it as "Hello, Good Morning!!!". Handle exceptions that can be generated during the I/O operations. Do not show the success message on the main exception handling block (write inside the else block).

````python
# write code here
def write_to_file():
    try:
        file = open("output.txt", "w")
        file.write("Hello, Good Morning!!!")
        file.close()

    except FileNotFoundError as e:
        print("FileNotFoundError:", e)

    except PermissionError as e:
        print("PermissionError:", e)

    except IOError as e:
        print("IOError:", e)

    else:
        print("Data written successfully.")

    finally:
        print("Execution completed.")


write_to_file()
````

### `Q-4`: Number game program.

Write a number game program. Ask the user to enter a number. If the number is greater than number to be guessed, raise a **ValueTooLarge** exception. If the value is smaller the number to be guessed the, raise a **ValueTooSmall** exception and prompt the user to enter again. Quit the program only when the user enters the correct number. Also raise **GuessError** if user guess a number less than 1.

````python
# Write code here
# Custom Exceptions
class ValueTooLarge(Exception):
    pass


class ValueTooSmall(Exception):
    pass


class GuessError(Exception):
    pass


def number_game(secret_number: int) -> None:
    """
    Number guessing game with custom exception handling.
    """
    print("Welcome to the Number Guessing Game!")

    while True:
        try:
            guess = int(input("Enter your guess: "))

            if guess < 1:
                raise GuessError("Guess must be greater than or equal to 1.")

            if guess > secret_number:
                raise ValueTooLarge("Your guess is too large.")

            if guess < secret_number:
                raise ValueTooSmall("Your guess is too small.")

            print("Congratulations! You guessed the correct number.")
            break

        except ValueTooLarge as e:
            print(e)

        except ValueTooSmall as e:
            print(e)

        except GuessError as e:
            print(e)

        except ValueError:
            print("Invalid input! Please enter a valid integer.")


number_game(secret_number=7)
````

### `Q-5:` Cast vote

Write a program that validate name and age as entered by the user to determine whether the person can cast vote or not. To handle the age, create **InvalidAge** exception and for name, create **InvalidName** exception. The name will be invalid when the string will be empty or name has only one word.

Example 1:

Input:

```bash
Enter the name:               goransh singh
Enter the age: 25
```

Output:

```bash
Goransh Singh  Congratulation !!! You can vote.
```

````python

class InvalidName(Exception):
    pass


class InvalidAge(Exception):
    pass


def validate_voter():
    try:
        name = input("Enter the name: ").strip()

        if not name:
            raise InvalidName("Name cannot be empty.")

        words = name.split()

        if len(words) < 2:
            raise InvalidName("Name must contain at least two words.")

        formatted_name = " ".join(word.capitalize() for word in words)

        age_input = input("Enter the age: ")

        try:
            age = int(age_input)
        except ValueError:
            raise InvalidAge("Age must be a valid integer.")

        if age < 0:
            raise InvalidAge("Age cannot be negative.")

        if age < 18:
            raise InvalidAge("You are not eligible to vote yet.")

        print(f"{formatted_name}  Congratulation !!! You can vote.")

    except InvalidName as e:
        print("InvalidName:", e)

    except InvalidAge as e:
        print("InvalidAge:", e)



validate_voter()
````

23### `Q-6`: Write a python function which infinitely prints natural numbers in a single line. Raise the **StopIteration** exception after displaying first 20 numnbers to exit from the program.

````python
# Write code here
def print_natural_numbers(n: int):
  for i in range(1, n+1):
    if i == 20:
      raise StopIteration
    else:
      print(i)
````

````python
print_natural_numbers(20)
````
