````python
# Write code here
# let's write something in file first

def write_in_file(filename: str, data: list):
    with open(filename, 'w') as f:
        for line in data:
            f.write(line)

l = ['hye\n', 'my name is pranshul\n', 'i live in delhi\n', 'i am learning ML\n']
write_in_file('demo.txt', l)


def get_final_line(filename):
    
    # open the file 
    with open(filename, 'r') as f:
        # create a list for each line in the file
        data = f.readlines()
        # fetch the last line using negative indexing and return it.
        return data[-1]

print(get_final_line('demo.txt'))
````

###`Q-4:` Create line wise reverse of a file
Write a function which takes two arguments: the names of the input file (to be read from) and the output file (which will be created).

For example, if a file looks like
 ```
abc def
ghi jkl
```
then the output file will be
```
fed cba
lkj ihg
```
**Notice**: The newline remains at the end of the string, while the rest of the characters are all reversed.

````python
# write code here
def reverse_content(input_file, output_file):

    with open(input_file, 'r') as f:
        data = f.readlines()

    with open(output_file, 'w') as f:
        for line in data:
            f.write(line.strip()[::-1] + '\n')

reverse_content('demo.txt', 'rev.txt')
````
