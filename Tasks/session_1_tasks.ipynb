### Q2:- Write a program that will convert celsius value to fahrenheit.

````python
# Write your code here
celsius = float(input('Enter temperature in celsius: '))
fahrenheit = (celsius * 9/5) + 32
print('Temperature in Fahrenheit is', fahrenheit)
````

### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here
x1 = int(input("Enter x1 coordinate: "))
x2 = int(input('Enter x2 coordinate: '))
y1 = int(input('Enter y1 coordinate: '))
y2 = int(input('Enter y2 coordinate: '))

distance = (((x2 - x1)**2) + ((y2 - y1)**2))**0.5
print('Distance between two coordinates is', distance)
````

### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python
# Write your code here
principle = float(input('Enter principle amount: '))
rate = float(input('Enter rate of interest: '))
time = float(input('Enter time period: '))

interest = (principle * rate * time)/100
print('Simple interest is', interest)
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
heads = int(input('Enter number of heads: '))
legs = int(input('Enter number of legs: '))

for i in range(heads + 1):
  j = heads - i
  if (2 * i + 4 * j) == legs:
    print('Number of dogs are', j)
    print('Number of chickens are', i)
    break
else:
  print('No solution')


````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here
n = int(input("Enter n: "))
total = 0
for i in range(1, n+1):
  total += i**2
print('Sum of squares of first', n, 'natural numbers is', total)

# second method
sum_of_squared = n * (n + 1) * (2 * n + 1) / 6
print(sum_of_squared)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
first_term = int(input('Enter first term: '))
second_term = int(input('Enter second term: '))
n = int(input('Enter n: '))

d = second_term - first_term
nth_term = first_term + (n - 1) * d
print('Nth term of the series is', nth_term)
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here
H = int(input('Enter height of the milk tank: '))
L = int(input('Enter length of the milk tank: '))
B = int(input('Enter breadth of the milk tank: '))

h = int(input('Enter height of the glass: '))
r = int(input('Enter radius of the glass: '))

volume_of_milk_tank = H * L * B
volume_of_glass = 3.14 * r**2 * h
print('Number of glasses of milk can be obtained is', volume_of_milk_tank // volume_of_glass)


````
