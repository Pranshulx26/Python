###`Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
Output : [(5, 6, 7, 8), (6, 10), (7, 13)]
```

````python
# write your code here
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]

ans = []
res = test_list[0]
for i in range(1, len(test_list)):
    if res[0] == test_list[i][0]:
        res += (test_list[i][1],)
    else:
        ans.append(res)
        res = test_list[i]
ans.append(res)
print(ans)


````

###`Q2:` Multiply Adjacent elements (both side) and take sum of right and lest side multiplication result.


For eg.
```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication :

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
# write your code here
test_list = (1, 5, 7, 8, 10)
left_prod = []
right_prod = []
for i in range(len(test_list)):
  if i == 0 and i+1 < len(test_list):
    left_prod.append(0)
    right_prod.append(test_list[i] * test_list[i+1])
  elif i == len(test_list)-1 and i-1 != 0:
    right_prod.append(0)
    left_prod.append(test_list[i] * test_list[i-1])
  else:
    left_prod.append(test_list[i-1]*test_list[i])
    right_prod.append(test_list[i]*test_list[i+1])

final_result = []
for i, j in zip(left_prod, right_prod):
  final_result.append(i+j)
print(final_result)
````

###`Q4`: Count no of tuples, list and set from a list
```
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

```
`Output:`

```
List-2
Set-2
Tuples-1
```

````python
# write your code here
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

tuple_count = 0
list_count = 0
set_count = 0

for item in list1:
  if isinstance(item, tuple):
    tuple_count += 1
  elif isinstance(item, list):
    list_count += 1
  elif isinstance(item, set):
    set_count += 1
print(tuple_count)
print(list_count)
print(set_count)
````

```

```

### `Q3`: Convert List to List of dictionaries. Given list values and keys list, convert these values to key value pairs in form of list of dictionaries.

**Example 1:**

Input:
```bash
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'DataScience', 'id': 3}, {'name': 'is', 'id': 8}]
```

**Example 2:**

Input:
```bash
test_list = ["CampusX", 10]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'CampusX', 'id': 10}]
```

````python
# write your code here
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]
i = 0
res = []
while i < len(test_list):
  d = dict()
  d[key_list[0]] = test_list[i]
  d[key_list[1]] = test_list[i+1]
  res.append(d)
  i += 2
print(res)
````

### `Q4`: Convert a list of Tuples into Dictionary.

**Example 1:**

Input:
```bash
[("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
```

Output:
```bash
{'akash': [10], 'gaurav': [12], 'anand': [14], 'suraj': [20], 'akhil': [25], 'ashish': [30]}
```

**Example 2:**

Input:
```bash
[('A', 1), ('B', 2), ('C', 3)]
```

Output:
```bash
{'A': [1], 'B': [2], 'C': [3]}
```

````python
# write your code here
lst = [("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
d = dict()
for item in lst:
  name, age = item
  if name not in d:
    d[name] = [age]
  else:
    d[name].append(age)
print(d)
````

### `Q5`: Sort Dictionary key and values List.

**Example 1:**

Input:

```bash
{'c': [3], 'b': [12, 10], 'a': [19, 4]}
```

Output:

```bash
{'a': [4, 19], 'b': [10, 12], 'c': [3]}
```

**Example 2:**

Input:

```bash
{'c': [10, 34, 3]}
```

Output:

```bash
{'c': [3, 10, 34]}
```

````python
# write your code here

d = {'c': [3], 'b': [12, 10], 'a': [19, 4]}

result = {k: sorted(v) for k, v in sorted(d.items())}

print(result)
````
