### **`Problem-1:`** Write a Python function that takes a list and returns a new list with unique elements of the first list.

**Exercise 1:**

Input:

```bash
[1,2,3,3,3,3,4,5]
```

Output:

```bash
[1, 2, 3, 4, 5]
```

````python
from typing import List
def unique_list(lst: List) -> List:
  """
  Create a new list with unique elements from the input list.

  Args:
    lst (List): A list of elements.

  Returns:
    A new list with unique elements from the input list.
  """
  temp = []
  for item in lst:
    if item not in temp:
      temp.append(item)
  return temp

unique_list([1,2,3,3,3,3,4,5])
````

### **`Problem-2:`** Write a Python function that accepts a hyphen-separated sequence of words as parameter and returns the words in a hyphen-separated sequence after sorting them alphabetically.

**Example 1:**

Input:
```bash
green-red-yellow-black-white
```

Output:
```bash
black-green-red-white-yellow
```

````python
# Write code here
def sort_words(words: str) -> str:
  """
  Sort a hyphen-separated sequence of words alphabetically.
  Args:
    words (str): A hyphen-separated sequence of words.
  Returns:
    A hyphen-separated sequence of words after sorting them alphabetically.
  """
  return '-'.join(sorted(words.split('-')))

sort_words('green-red-yellow-black-white')
````

### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
# Write code here
def count_upper_lower(string: str) -> None:
  """
  Count the number of upper case and lower case letters in a string.

  Args:
    string (str): A string.

  Returns:
    None
  """
  upper = 0
  lower = 0
  for char in string:
    if char.isupper():
      upper += 1
    elif char.islower():
      lower += 1
  print(f'No. of Upper case characters : {upper}')
  print(f'No. of Lower case characters : {lower}')
count_upper_lower('CampusX is an Online Mentorship Program fOr EnginEering studentS.')
````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
# Write code here
def is_perfect_number(number: int) -> bool:
  """
  Check whether a number is perfect or not.

  Args:
    number (int): a number to validate if it's perfect or not

  Returns:
    bool: True if the number is perfect, False otherwise
  """
  sum = 0
  for i in range(1, number+1):
    if number % i == 0:
      sum += i
  if sum // 2 == number:
    return True
  return False

is_perfect_number(28)
````

### **`Problem-6:`** Write a Python function to concatenate any no of dictionaries to create a new one.

```
Sample Dictionary :
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}
```

````python
# Write code here
from typing import Dict, Any


def merge_dictionaries(*dicts: Dict[Any, Any]) -> Dict[Any, Any]:
    """
    Merge any number of dictionaries into a new dictionary.

    Args:
        *dicts: Variable number of dictionary arguments.

    Returns:
        Dict[Any, Any]: A new merged dictionary.

    Raises:
        TypeError: If any argument is not a dictionary.
    """
    merged: Dict[Any, Any] = {}

    for d in dicts:
        if not isinstance(d, dict):
            raise TypeError(f"Expected dict, got {type(d).__name__}")
        merged.update(d)

    return merged


dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}

result = merge_dictionaries(dic1, dic2, dic3)
print(result)


````

`Problem-7` Write a python function that accepts a string as input and returns the word with most occurence.

```
Input:
hello how are you i am fine thank you
```

```
Output
you -> 2
```

````python
# Write code here
def most_frequent_word(string: str) -> str:
  """
  Find the word with the most occurrences in a string.

  Args:
    string (str): A string.

  Returns:
    str: The word with the most occurrences in the string.
  """
  words = string.split()
  d = {}
  for word in words:
    if word in d:
      d[word] += 1
    else:
      d[word] = 1

  return sorted(d.items(), key=lambda x: x[1], reverse=True)[0]

most_frequent_word('hello how are you i am fine thank you')
````

`Problem-8` Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
Input:
[13,42,15,37,22,39,41,50]
```

```
Output:
{11-20:2,21-30:1,31-40:2,41-50:3}
```

````python
def histogram_bins(numbers: list[int]) -> dict[str, int]:
    """
    Create a histogram with bin size 10.

    Args:
        numbers (list[int]): List of integers.

    Returns:
        dict[str, int]: Histogram dictionary with bin ranges as keys.
    """
    histogram = {}

    for num in numbers:
        if num <= 0:
            continue

        lower = ((num - 1) // 10) * 10 + 1
        upper = lower + 9
        key = f"{lower}-{upper}"

        histogram[key] = histogram.get(key, 0) + 1

    return dict(sorted(histogram.items()))

histogram_bins([13,42,15,37,22,39,41,50])
````

`Problem-9` Write a python function that accepts a list of 2D co-ordinates and a query point, and then finds the the co-ordinate which is closest in terms of distance from the query point.

```
List of Coordinates
[(1,1),(2,2),(3,3),(4,4)]
Query Point
(0,0)
```

```
Output
Nearest to (0,0) is (1,1)
```

````python
# Write code here
from typing import List, Tuple
import math
def closest_point(coordinates: List[Tuple], query: Tuple) -> Tuple:
  """
  Find the closest point to a query point in a list of 2D coordinates.
  Args:
    coordinates (List[Tuple]): A list of 2D coordinates.
    query (Tuple): A 2D query point.
  Returns:
    Tuple: The closest point to the query point.
  """
  min_distance = math.inf
  closest_point = None
  for point in coordinates:
    distance = math.sqrt((point[0] - query[0])**2 + (point[1] - query[1])**2)
    if distance < min_distance:
      min_distance = distance
      closest_point = point
  return closest_point

closest_point([(1,1),(2,2),(3,3),(4,4)], (0,0))
````

`Problem-10`:Write a python program that receives a list of strings and performs bag of word operation on those strings

https://en.wikipedia.org/wiki/Bag-of-words_model

````python
# Write code here
from collections import Counter
def bag_of_words(strings: list) -> dict:
  """
  Perform bag of word operation on a list of strings.

  Args:
    strings (list): A list of strings.

  Returns:
    dict: A dictionary containing the bag of words.
  """
  words = []
  for string in strings:
    words += string.split()
  return Counter(words)

bag_of_words(['hello how are you', 'i am fine thank you'])
````
