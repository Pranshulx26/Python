###`Q-4:` Your are given an array which is havig some nan value. You job is to fill those nan values with most common element in the array.
```
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

```

````python
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])
arr
````

````python
d = {}
for i in np.nditer(arr):
    i = float(i)
    if i not in d:
        d[i] = 1 
    else:
        d[i] += 1


high_freq_element = sorted(d.items(), key = lambda item: item[1], reverse=True)[0][0]

np.where(np.isnan(arr), high_freq_element, arr)
````

````python
# code here
import numpy as np

arr = np.array([[1, 2, np.nan], [4, 2, 6], [np.nan, np.nan, 5]])

# 1. Get all non-NaN elements
non_nan_elements = arr[~np.isnan(arr)]

# 2. Find the most common element (mode)
vals, counts = np.unique(non_nan_elements, return_counts=True)
most_common = vals[np.argmax(counts)]

# 3. Fill the NaN values with the most common element
arr[np.isnan(arr)] = most_common

print(arr)
# Output:
# [[1. 2. 2.]
#  [4. 2. 6.]
#  [2. 2. 5.]]
````

### `Q-9:` Write a program implement Boradcasting Rule to check if two array can be added or not.
Given tuples of shapes.
```
shape of a- (3,2,2)
shape of b- (2,2)

check_broadcast(a, b) -> return Boolean (True if can broadcasted, False other wise.)
```

````python
# code here
def check_broadcast(a_shape, b_shape):
    # Align shapes by padding the shorter one with 1s on the left
    len_a, len_b = len(a_shape), len(b_shape)
    max_len = max(len_a, len_b)
    
    # Pad with 1s: (2,2) becomes (1,2,2) to match (3,2,2)
    shape1 = (1,) * (max_len - len_a) + a_shape
    shape2 = (1,) * (max_len - len_b) + b_shape
    
    # Check compatibility from right to left
    for dim1, dim2 in zip(shape1, shape2):
        if dim1 != dim2 and dim1 != 1 and dim2 != 1:
            return False
            
    return True


````
