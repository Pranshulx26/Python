Namespaces

A namespace is a space that holds names(identifiers). Programatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values).

There are 4 types of namespaces:

- Builtin Namespace
- Global Namespace
- Encloding Namespace
- Local Namespace

Scope and LEGB Rule

A scope is a textual region of a python program where a namespace is directly accessible

The interpreter searches for a name from the inside out, looking in the local, enclosing, global and finally the built-in scope. if the interpreter does not find the name in any of these locations, then python raises a NameError exception

```python
# local and global 

# global variable
a = 2 

def temp():
    # local variable
    b = 3 
    print(b)

temp()
print(a) 
```

```python
# global variable
a = 2 

def temp():
    # local variable
    a = 3 
    print(a)

temp()
print(a) 
```

```python
# local and global -> editing global

# global variable
a = 2 

def temp():
    # local variable
    a += 1
    print(a)

temp()
print(a) 

# we can only view global variables not change them by default
```

```python
# global variable
a = 2 

def temp():
    global a 
    # local variable
    a += 3
    print(a)

temp()
print(a) 
```

```python
# global variable
a = 2 

def temp(a):
    global a 
    # local variable
    a += 3
    print(a)

temp(a)
print(a) 
```

```python
# local and global -> global created inside local

def temp():
    # creating global variable in function
    global a 
    a = 1
    print(a)

temp()
print(a)
```

```python
# local and global -> function parameter is local
def temp(z):
    # local var 
    print(z)

a = 5
temp(5)
print(a)
```

```python
# Enclosing scope

def outer():
    def inner():
        print('inner function')
    inner()
    print('outer function') 
outer()
print('main program')

```

```python
# LEGB - "a" present in local will print 4
def outer():
    a = 3
    def inner():
        a = 4
        print(a) 
        
    inner()
    print('outer function') 
a = 1 
outer()
print('main program')
```

```python
# LEGB - "a" not present in local but present in enclosing will print 3
def outer():
    a = 3
    def inner():
        print(a) 
        
    inner()
    print('outer function') 
a = 1 
outer()
print('main program')
```

```python
# LEGB - "a" not present in local and enclosing but present in global will print 1
def outer():
    def inner():
        print(a) 
        
    inner()
    print('outer function') 
a = 1 
outer()
print('main program')
```

```python
# nonlocal keyword
def outer():
    a = 1 
    def inner():
        a += 1
        print(a) 
        
    inner()
    print('outer function') 
a = 1 
outer()
print('main program')
```

```python
# nonlocal keyword
def outer():
    a = 1 
    def inner():
        nonlocal a 
        a += 1
        print('inner', a) 
        
    inner()
    print('outer', a) 
a = 1 
outer()
print('main program')
```

Decorators

A decorators in python is a function that receives another function as input and adds some functionality(decoration) to and it and return it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python

- build in decorators like @staticmethod, @classmethod, @abstractmethod and @property etc
- user defined decorators that we programmers can create according to our needs.


```python
# Python are 1st class function
def func():
    print('hello')

del func 
func()
```

```python
def modify(func, num):
    return func(num)

def square(num):
    return num**2 

modify(square, 2)
```

```python
# simple example 

def my_decorator(func):
    def wrapper():
        print('*********************************')
        func()
        print('*********************************')

    return wrapper 

def hello():
    print('hello') 

def display():
    print('hello pranshul') 

a = my_decorator(hello)
a()

my_decorator(display)()
```

```python
# inner function access kar sakta h outer function ke variables ko even though outer 
# function has already return and removed from memory it's a special case called closure
```

```python
# better syntax 

def my_decorator(func):
    def wrapper():
        print('*********************************')
        func()
        print('*********************************')

    return wrapper 

@my_decorator
def hello():
    print('hello') 

hello()
```

```python
# anything meaningful?
import time 

def timer(func):
    def wrapper(*args):
        start = time.time()
        func(*args)
        end = time.time()
        print(f'time taken to execute the function {func.__name__}: ', round(end - start, 2), 'seconds')
    return wrapper

@timer
def func():
    for _ in range(100000000):
        continue 

@timer
def square(num):
    time.sleep(1)
    print(num**2)
#func()
square(2)


```

```python
# A big problem 


def sanity_check(data_type):
    def outer_wrapper(func):
        def inner_wrapper(*args):
            if type(*args) == data_type:
                func(*args)
            else:
                raise TypeError('Ye data type nai chalega')
        return inner_wrapper
    return outer_wrapper
    
@sanity_check(int)
def square(num):
    print(num**2)
square(2)
square('hehe')
```
