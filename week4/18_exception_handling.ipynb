Exceptions

If things go wrong during the execution of the program(runtime). It generally happens when something unforeseen has happened.

- Exceptions are raised by python runtime
- You have to take is on the fly.

Examples

- Memory overflow
- Divide by 0 -> Logical error
- Database Error

```python
# Why is it important to handle exception 
# -> USER EXPERIENCE AND SECURITY
# How to handle exceptions
# -> Try except block
```

```python
# let's create a file
with open('sample.txt', 'w') as f:
    f.write('hello world')
```

```python
# try catch demo 
try:
    with open('sample100.txt', 'r') as f:
        print(f.read())
except:
    print('sorry file not found')
```

```python
# catching specific exception

try:
    f = open('sample.txt', 'r') 
    print(f.read())
    print('')
    print(5/1)
    l = [1, 2, 3]
    l[10]
except FileNotFoundError:
    print('file not found')
except NameError:
    print("Varialbe not defined")
except ZeroDivisionError:
    print('Can"t divide by 0')
except Exception as e:
    print(e)
```

```python
# else
try:
    f = open('sample.txt', 'r')
except FileNotFoundError:
    print('file nai mili')
except Exception:
    print('kuch to lafda hai')
else:
    print(f.read())
```

```python
# else
try:
    f = open('samplesfa.txt', 'r')
except FileNotFoundError:
    print('file nai mili')
except Exception:
    print('kuch to lafda hai')
else:
    print(f.read())
```

```python
class Bank:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount < 0:
            raise Exception('amount cannot be -ve')
        if self.balance < amount:
            raise Exception('paise nhi h tere paas')
        self.balance = self.balance - amount

obj = Bank(10000)
try:
    obj.withdraw(-4000)
except Exception as e:
    print(e)
else:
    print(obj.balance)
```

```python
# creating custom exceptions

class MyException(Exception):
    def __init__(self, message):
        print(message)

class Bank:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount < 0:
            raise MyException('amount cannot be -ve')
        if self.balance < amount:
            raise MyException('paise nhi h tere paas')
        self.balance = self.balance - amount

obj = Bank(10000)
try:
    obj.withdraw(-4000)
except MyException as e:
    pass
else:
    print(obj.balance)
```

```python
# why should we use custom classes?
# we are creating login/registration class
# ask user name/email/password/device signature (windows / MAC)

class SecurityError(Exception):
    def __init__(self, message):
        print(message)

    def logout(self):
        print('logout')
class Google:

    def __init__(self, name, email, password, device):
        self.name = name 
        self.email = email
        self.password = password
        self.device = device

    def login(self, email, password, device):
        if device != self.device:
            raise SecurityError('bhai teri to lag gayi')
        if email == self.email and password == self.password:
            print('welcome')
        else:
            print('login error')

obj = Google('pranshul', 'pranshul@gmail.com', '1234', 'android')

try:
    obj.login('pranshul@gmail.com', '1234', 'windows')
except SecurityError as e:
    e.logout()
else:
    print(obj.name)
finally:
    print('database connection closed')
```
