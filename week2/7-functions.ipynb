Let's create a function(with docstring)

```python
from typing import Literal


def is_even(number: int) -> Literal["Even", "Odd"]:
    """
    Determine whether a given integer is even or odd.

    Args:
        number (int): The integer to evaluate.

    Returns:
        Literal["Even", "Odd"]: "Even" if the number is even,
        otherwise "Odd".

    Raises:
        TypeError: If `number` is not an integer.
    """
    if not isinstance(number, int):
        raise TypeError(
            f"Expected type 'int', but got '{type(number).__name__}' instead."
        )

    return "Even" if number % 2 == 0 else "Odd"


print(is_even(10))
print(is_even(11))
```

```python

from multiprocessing import Value
from numbers import Real 
from typing import Union 

def power(a: Real = 1, b: Real = 1) -> Union[int, float]:
    """
    Compute a raised to the power of b (a ** b).

    Args: 
        a (Real): Base value (int or float). Defaults to 1.
        b (Real): Exponent value (int or float). Defaults to 1.

    Returns:
        Union[int, float]: Result of exponentiation

    Raises:
        TypeError: If 'a' or 'b' is not a real number.
        ValueError: If undefined mathematical case occurs (eg. 0 ** negative).

    """

    if not isinstance(a, Real):
        raise TypeError(f"Base 'a' must be a real number, got {type(a).__name__}")
    if not isinstance(b, Real):
        raise TypeError(f"Base 'b' must be a real number, got {type(b).__name__}")
    
    # handling undefined case
    if a == 0 and b < 0:
        raise ValueError('0 cannot be raised to a negative power.')

    result = a ** b 
    return result 

print(power(2, 3))
print(power(2))
print(power())
```

*args and **kwargs

are special python keywords that are used to pass the variable length of arguments to a function.

```python
# *args
# allows us to pass a variaable number of non-keyword arguments to a function.

def multiply(*args):
    prod = 1 
    for i in args:
        prod *= i
    return prod 

multiply(2, 3, 5, 10)
```

```python
print(is_even.__doc__)
```

```python
# **kwargs
# **kwargs allows us to pass any number of keyword arguments.
# keyword arguments mean that they contain a key-value pair, like a python dictionary
```

```python
def display(**kwargs):
    
    for key, value in kwargs.items():
        print(key, '->', value)

display(india='delhi', srilanka='colombo', nepal='katmando')
```

Points to remember while using *args -> **kwargs

- order of arguments matter(normal -> *args -> **kwargs)
- The worlds *args and **kwargs are only a convention, you can use any name of your choice.

Lambda Function

A lambda function is a small anonymous function.

A lambda function can take any number of arguments, but can only have one expression.

```python
# x -> x**2
a = lambda x: x**2
a(2)
```

```python
# x, y -> x + y
b = lambda x, y: x + y 
b(5, 2)
```

Difference between lambda vs normal function

- No name
- Lambda has no return value (infact, returns a function)
- lambda is written in 1 line
- not reusable

The why use lambda function?
They are used with HOF(higher order function)

```python
# check if a string has 'a' 
check_a = lambda string: 'a' in string 
check_a('pranshul')
```

```python
# odd or even
a = lambda x: 'even' if x % 2 == 0 else 'odd'
a(5)
```

Higher order functions

- A function that returns a function
- A function that receives another function as an input

```python
# Example 

def square(x):
    return x ** 2 

def cube(x):
    return x ** 3 

# Higher order function
def transform(f, L):
    output = []
    for i in L:
        output.append(f(i))
    print(output)

L = [1,2,3,4,5]
transform(square, L)
transform(lambda x: x**3, L)
```

Map()

```python
# Square the items of a List
list(map(lambda x: x**2, [1, 2, 3, 4, 5]))
```

```python
# odd/even labelling of list item
l = [1, 2, 3, 4, 5]
list(map(lambda x: 'even' if x % 2 == 0 else 'odd', l))
```

Filter

```python
# numbers greater than 5
l = [1, 2, 3, 4, 5, 6, 7, 8]
list(filter(lambda x: x>5, l))
```

Reduce()

```python
# sum of all item 
import functools

functools.reduce(lambda x,y : x + y, [1, 2, 3, 4, 5])
```

```python
# find min 
functools.reduce(lambda x, y : x if x < y else y, [23, 42, 1, 2, 4])
```
