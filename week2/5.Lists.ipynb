What are Lists

List is a data type where you can store multiple items under 1 name. More technically lists acts list dynamic arrays which means you can add more items on the fly.

Array Vs Lists (Refrential Arrays)

- Fixed vs Dynamic Size
- Convenience -> Hetrogeneous
- Speed of Execution
- Memory

```python
L = [1, 2, 3]
print(id(L))
print(id(L[0]))
print(id(L[1]))
print(id(L[2]))
print(id(1))
print(id(2))
print(id(3))
```

Creating a List

```python
# Empty
print([])
# 1D
print([1, 2, 3, 4])
# 2D
print([[1,2], [3,4]])
# 3D
print([[[1,2], [3, 4]], [[5,6], [7,8]]])
# Hetrogenous 
print(['hello', 1, 1.2, True, {1, 2, 3}, {'hye': 1}, (1, 2, 3)])
# Using Type Conversion
print(list('hello'))
```

Accessing items from a List

```python
# Indexing
l = [1, 2, 3, 4]
print(l)
# positive indexing
print('positive indexing examples: ')
print(l[0])
print(l[1])
print(l[2])
print(l[3])
# negative indexing
print('negative index examples: ')
print(l[-1])
print(l[-2])
print(l[-3])
print(l[-4])
# Slicing
print('slicing in list: ')
print(l[:2])
print(l[:3])
print(l[1:3])
print(l[1:])
print(l[:4])

# reverse list
print('reverse list: ')
print(l[::-1])

# skip every other element
print('skip every other element in list: ')
print(l[::2])

```

```python
l_2d = [[1, 2], [3, 4]]
print(l_2d[0][1]) # 2

# fetch 3
print(l_2d[1][0])
```

List Comprehension

List comprehension provides a concise way of creating lists.

newlist = [expresion for item in iterable if condition == True]

Advantages of List Comprehension

- More time-efficient and space-efficient that loops
- Require fewer lines of code
- Transforms iterative statement into a formula

```python
# Add 1 to 10 numbers to a list
l = [i for i in range(1, 11)]
print(l)
```

```python
# Scalar multiplication on a vector
l = [1, 2, 3, 4]
# multiply each item in l by 2 
new_list = [i*2 for i in l]
print(new_list)
```

```python
# Add squares
l = [1, 2, 3, 4, 5]
[i**2 for i in l]
```

```python
# print all numbers divisible by 5 in the range of 1 to 50
[i for i in range(1, 51) if i%5==0]
```

```python
# find languages which start with letter p
languages = ['java', 'python', 'php', 'c', 'javascript']
[lang for lang in languages if lang.startswith('p')]
```

```python
# Nested if with List comprehension
basket = ['apple', 'guava', 'cherry', 'banana']
my_fruits = ['apple', 'kiwi', 'grapes', 'banana']

# add new list from my_fruits and items if the fruit exists in basket and also starts with 'a'
[fruit for fruit in my_fruits if fruit in basket if fruit.startswith('a')]
```

```python
# print a (3,3) matrix using list comprehension -> Nested List Comprehension
[[i * j for i in range(1, 4)] for j in range(1, 4)]
```

```python
# Cartesian products -> List comprehension on 2 lists together
l1 = [1, 2, 3, 4]
l2 = [5, 6, 7, 8]

[i*j for i in l1 for j in l2]
```

Zip
The zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

If the passed iterators have different lengths, the iterator with the least items decides the length of the new iterator.

```python
# Write a program to add items of 2 lists indexwise

l1 = [1, 2, 3, 4]
l2 = [-1, -2, -3, 4]

list(zip(l1, l2))

[i+j for i,j in zip(l1, l2)]
```

```python
# Create 2 lists from a given list where
# 1st list will contain all the odd numbers from the original list and 
# 2nd one will contain all the even numbers

lst = [1, 2, 3, 4, 5, 6]
odd = []
even = []
for item in lst:
    if item % 2 == 0:
        even.append(item)
    else:
        odd.append(item)
print(odd)
print(even)
```

```python
# Write a program to check if a list is in ascending order or not?
l = [1, 2, 3, 4, 5]
flag = True 
for i in range(1, len(l)):
    if l[i-1] > l[i]:
        flag = False
        break
if flag:
    print('list is in ascending order')
else:
    print('list is not in ascending order')
```
