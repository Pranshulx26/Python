Tuples 

A tuple in python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assignmed whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple cannot be changed in any way once it is created.

Characterstics:
- Ordered
- Unchangeble
- Allows duplicates

Creating Tuples

```python
# empty 
t1 = ()
print(type(t1))
# create a tuple with a single item
t2 = (2)
print(type(t2))
t3 = (2,)
print(type(t3))
# homogeneous
t4 = (1, 2, 3, 4)
print(t4)
# hetro 
t5 = (1 , True, print, dict)
print(t5)
# using type conversion
t6 = tuple('hello')
print(t6)
```

Accessing items
- Indexing
- Slicing

```python
t = (1, 2, 3, 4, 5)
# positive indexing
print(t[0])
print(t[1])
print(t[2])
# negative indexing
print(t[-1])
# slicing
print(t[:2])
print(t[1:4])
print(t[1::2])
# reverse tuple
print(t[::-1])
```

```python
# 2d tuple
t2 = ((1,2), (3,4))
t2[1][1]
```

Editing items

```python
t = (1, 2, 3, 4)
t[1] = 1231 # immutable
```

Adding items

```python
print(t)
# no append, insert, extend
```

Deleting items

```python
# we can delete the whole tuple but not elements
t = (1, 2, 3)
print(t)
del t 
print(t)
```

```python
t = (1, 2, 3, 4)
del t[:3]
```

Operations on Tuples

```python
# + and *
t = (1, 2, 3, 4)
t2 = (5, 6, 7, 8)
print(t + t2)
print(t * 3)
# membership
print(1 in t)
# loops
for i in t:
    print(i)
```

Tuple Functions

```python
# len/ sum/ min/ max/ sorted
t = (1, 2, 3, 4)
print(len(t))
print(max(t))
print(min(t))
print(sum(t))
print(sorted(t, reverse=True))
```

```python
# count
t = (1, 1, 1, 2, 3, 4)
t.count(1)
```

```python
# index 
t = (1, 2, 3, 4, 5)
t.index(3)
```

```python
t.index(53)
```

Difference between Lists and Tuples
- Syntax
- Mutability
- Speed - Tuples are faster than list
- Memory - Tuples take less memeory than list
- Built in functionality - list has more functions
- Error prone - lists are more error prone
- Usability 

```python
import sys 

l = list(range(1000))
t = tuple(range(1000))

print('List size: ',sys.getsizeof(l))
print('Tuple size: ', sys.getsizeof(t))
```

```python
a = [1, 2, 3]
b = a 
a.append(4)
print(a)
print(b)
```

```python
a = (1, 2, 3)
b = a 
a = a + (4,)
print(a)
print(b) # no change
```

Sets

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference etc.

Characterstics:
- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

Creating Sets

```python
s = set()
print(type(s))
```

```python
# empty
s = set()
print(type(s))
print(s)

# 1D and 2D
set_1d = {1, 2, 3, 4}
print(set_1d)
print(type(set_1d))

set_2d = {{1, 2, 3}, {4, 5, 6}} # error - sets are mutalbe and set cannot contain mutable objects
print(set_2d)
```

```python
# homo and hetro
s1 = {1, 2, 3, 4}
print(s1)

# hetro
s2 = {1, True, 'hello', 4.5}
print(s2) # only three items because True and 1 are same
```

```python
# using type conversion

s3 = set(tuple([1, 2, 3, 4, 5]))
print(s3)
```

```python
# duplicates not allowed
s4 = {1, 2, 2, 2, 2, 3, 3, 3, 4, 4}
print(s4)
```

```python
# set can't have mutalbe items 
# lists are mutalble
s5 = {[1, 2, 3], {1, 2, 3}}
```

```python
s = {1, 2, 3}
s2 = {3, 2, 1}
print(s == s2)
```

Accessing Items

```python
s1 = {1, 2, 3, 4}
s1[0] # can't do indexing when unordered
```

Editing items

```python
s1 = {1, 2, 3, 4}
s1[0] = 1000
```

Adding items

```python
s = {1, 2, 3, 4}

# add
s.add(5)
print(s)

# update
s.update([5, 6, 7, 8])
print(s)
```

Deleting items

```python
s = {1, 2, 3, 4, 5}

# del
del s # can't delete single items or multiple because item assignment not possible

s = {1, 2, 3, 4, 5}
# discard
s.discard(5)
print(s)
s.discard(5000)
print(s)

# remove
s.remove(4)
print(s)
s.remove(4000)
print(s) # error


```

```python
s = {1, 2, 3}
# pop
s.pop() # deletes item randomly
print(s) 

# clear
s.clear() # empty set 
print(s)
```

Set Opeartions

```python
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

# union(|)
print(s1 | s2 )

# intersection
print(s1 & s2)

# Difference(-)
print(s1 - s2) # items of s1 not in s2

# Symmetric Difference (^) # removing commons
print(s1 ^ s2)

# Membership Test
print(1 in s1) 
print(1 not in s1) 

# Loops 
for i in s1:
    print(i)


```

Set Functions

```python
# len/ sum/ min/ max/ sorted
s = {1, 2, 3, 4}
print(len(s))
print(sum(s))
print(min(s))
print(max(s))
print(sorted(s))
```

```python
# union/update
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

s1.union(s2)
```

```python
print(s1)
s1.update(s2) # permanent changes in s1
print(s1)
print(s2)
```

```python
# Intersection/intersection_update
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

s1.intersection(s2) 
s1.intersection_update(s2)
print(s1)
```

```python
# difference/ difference_update
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

s1.difference(s2) 
s1.difference_update(s2)
print(s1)
```

```python
# symmetric_difference/ symmetric_difference_update
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

s1.symmetric_difference(s2) 
s1.symmetric_difference_update(s2)
print(s1)
```

```python
# isdisjoint 

s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

s1.isdisjoint(s2) # s1 and s2 have common
```

```python
# issubset 
s1 = {1, 2, 3}
s2 = {2, 3}
s2.issubset(s1)
```

```python
# issuperset
s1.issuperset(s2)
```

```python
# copy
s = s1.copy() # shallow copy # address will be different
print(s)
```

Dictionary

Dictionary in Python is a collection of keys values, used to store data values like a map, which unlike other data types which hold only a single value as an element.

In some language it is known as map or assosiative arrays.

dict = {'name': 'pranshul', 'age': 33}

Characterstics:

- Mutable
- Indexing has no meaning
- Keys can't be duplicated
- Kyes can't be mutable items

```python
# empty dictionary
d = {}
print(d)
print(type(d))
# 1D dictionary
d1 = {'name': 'pranshul'}
print(d1)
# with mixed keys
d2 = {(1, 2, 3): 1, 'hello': 'world'} # tuples are immutable
print(d2) 
# 2d dictionary
s = {
    'name': 'pranshul',
    'college': 'ims',
    'semester': '8th',
    'subjects': {
        'dsa': 90,
        'maths': 88
    }
}
print(s)

# using sequence and dict function

d4 = dict([(1, 1), (2,2), (3, 3), ('name', 'pranshul')])
print(d4)

# duplicate keys
d5 = {'name': 'pranshul', 'name': 'rahul'}
print(d5)

# mutable items as keys not allowed
d6 = {'name': 'nitish', [1]: 'al;sdfj'}
print(d6)
```

Dictionary Comprehension

{key: value for vars in iterable}

```python
# print 1st 10 numbers and their squares
{i: i**2 for i in range(1, 11)}
```

```python
# using existing dict
distances = {'delhi': 1000, 'mumbai': 2000}
{key: value*0.62 for key, value in distances.items() }
```

```python
# using zip
days = ['sunday', 'monday', 'tuesday', 'wednesday', 'thursday', 'friday', 'saturday']
temp_c = [30.5, 32.6, 31.8, 33.8, 32.3, 33.5, 36.5]
{i: j for i, j in zip(days, temp_c)}
```

```python
# using if condition
products = {"phone": 10, 'laptop': 0, 'charger': 32}
{key: value for key, value in products.items() if value > 0}
```

```python
# Nested Comprehension
# print tables of number from 2 and 4
# {
#     2:{1:2, 2:4, 3:6},
#     3:{1:3, 2;6...},
#     4:{1:4, 2:8, ...}
# }

{i: {j: i*j for j in range(1, 11)} for i in range(2, 5)}
```
