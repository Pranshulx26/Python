```python
number = int(input('Enter the number'))

digits = '0123456789'
result = ''
while number != 0:
    result = digits[number % 10] + result 
    number = number // 10 
print(result)


# TC: LOG(N)
```

```python
L = [1, 2, 3, 4]

sum = 0
for i in L:
    sum = sum + i 

product = 1
for i in L:
    product = product * i

print(sum, product)

# TC: O(N) + O(N) -> O(N)
```

```python
A = [1, 2, 3, 4]
B = [5, 6, 7, 8]
for i in A:
    for j in B:
        print(i, j)

# TC: O(N**2)
```

```python
A = [1, 2, 3, 4]
B = [5, 6, 7, 8]

for i in A:
    for j in B:
        for k in range(1000):
            print(i, j)

# TC: O(N**2)
```

```python
L = [1, 2, 3, 4, 5]

for i in range(len(L)//2):
    other = len(L) - i - 1
    temp = L[i]
    L[i] = L[other]
    L[other] = temp
print(L)

# TC: O(N)
```

```python
n = 10 
k = 0
for i in range(n//2, n):
    for j in range(2, n, pow(2, j)):
        k = k + n / 2
print(k)

# TC: N * LOG(N)
```

```python
n = 345

sum = 0
while n > 0:
    sum = sum + n % 10
    n = n // 10
print(sum) 

# TC: O(LOG(N))
```

```python
def fib(n):
    if n == 1 or n == 0:
        return 1
    else:
        return fib(n-1) + fib(n-2)
    
    
# TC: Exponentiation
```
