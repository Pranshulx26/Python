Write OOP classes to handle the following scenarios:

- A user can create and view 2D coordinates
- A user can find out the distance between 2 coordinates
- A user can find the distance of a coordinate from origin
- A user can check if a point lies on the line or not.
- A user can find the distance between a given 2D point and a given line

How ojbects access attributes

```python
class Point:

    def __init__(self, x, y):
        self.x = x 
        self.y = y 

    def __str__(self):
        return f"<{self.x}, {self.y}>"
    
    def euclidean_distance(self, other):
        return ((self.x - other.x)**2 + (self.y - other.y)**2)**0.5
    
    def distance_from_origin(self):
        return self.euclidean_distance(self, Point(0, 0))
    
class Line:
    def __init__(self, A, B, C):
        self.A = A 
        self.B = B 
        self.C = C 

    def __str__(self):
        return f'{self.A}x + {self.B}y + {self.C} = 0'
    
    def point_on_line(line, point):
        if line.A * point.x + line.B * point.y + line.C == 0:
            return 'lies on the line'
        else:
            return 'Does not lie on the line'
        
    def shortest_distance(line, point):
        numerator = abs(line.A * point.x + line.B * point.y + line.C)
        denomenator = (line.A ** 2 + line.B ** 2 )**0.5
        result = numerator / denomenator
        return result 
    
    
```

```python
p1 = Point(0, 0)
p2 = Point(1, 1)
# print(p1)
# print(p2)
p1.euclidean_distance(p2)
```

```python
l1 = Line(3, 4, 5)
print(l1)
```

```python
l1 = Line(1, 1, -2)
p1 = Point(1, 10)
print(l1)
print(p1)

l1.point_on_line(p1)
```

```python
l1 = Line(1, 1, -2)
p1 = Point(1, 10)
l1.shortest_distance(p1)
```

Reference Variables

- Reference variables hold the objects
- We can create objects without reference variable as well
- An object can have multiple reference variables
- Assigning a new reference variable to an existing object does not create a new object.

```python
# object without a reference
class Person:

    def __init__(self):
        self.name = 'pranshul'
        self.gender = 'male'
```

```python
p = Person() # p is storing the reference of the object
# p is not the object it's a container that holds the address of the object
```

```python
# multiple ref 
q = p 
print(id(p))
print(id(q))
```

```python
# change attribute value with the help of 2nd object
print(p.name)
print(q.name)
q.name = 'ankit'
print(q.name)
print(p.name)
```
