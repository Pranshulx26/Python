```python
# example
class Customer:

    def __init__(self, name, gender, address):
        self.name = name 
        self.gender = gender 
        self.address = address 

    def print_address(self):
        print(self.address.get_city(), self.address.pin, self.address.state)

    def edit_profile(self, new_name, new_city, new_pin, new_state):
        self.name = new_name 
        self.address.edit_address(new_city, new_pin, new_state)


class Address:
    
    def __init__(self, city, pincode, state):
        self.__city = city 
        self.pin = pincode
        self.state = state 

    def get_city(self):
        return self.__city

    def edit_address(self, new_city, new_pin, new_state):
        self.__city = new_city
        self.pin = new_pin 
        self.state = new_state

add1 = Address('gurgaon', 122011, 'Haryana')

cust = Customer('pranshul', 'male', add1) 
cust.print_address()
cust.edit_profile('ankit', 'mumbai', 11111, 'maharashtra')
cust.print_address()
# method example 

# what about private attribute 
```

Super Keyword

```python


class Phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor") 
        self.price = price 
        self.brand = brand 
        self.camera = camera

    def buy(self):
        print('Buying a phone')

class SmartPhone(Phone):

    def buy(self):
        print('Buying a smartphone')
        # syntax to call parent ka buy method
        super().buy() 

s = SmartPhone(20000, 'Apple', 12)

s.buy()
```

```python
# super -> constructor

class Phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor") 
        self.price = price 
        self.brand = brand 
        self.camera = camera

    def buy(self):
        print('Buying a phone')

class SmartPhone(Phone):

    def __init__(self, price, brand, camera, os, ram):
        super().__init__(price, brand, camera) 
        self.os = os 
        self.ram = ram 
        print('Inside smartphone constructor')
    def buy(self):
        print('Buying a smartphone')
        # syntax to call parent ka buy method
        super().buy() 

s = SmartPhone(20000, 'Apple', 12, 'android', 2)

print(s.os ) 
print(s.brand)
```

Inheritance Summary:

- A class can inherit from another class.
- Inheritance improves code reusabalility
- Constructor, attributes, methods get inherited to the child class
- The parent has no access to the child class
- Private properties of parent are not accessible directly in child class
- Child class can override the attributes or methods. This is called method overriding
- super() is an inbuild function which is used to invoke the parent class methods and constructor.

```python
class Parent:
    def __init__(self, num):
        self.__num = num 
    def get_num(self):
        return self.__num 

class Child(Parent):
    def __init__(self, num, val):
        super().__init__(num)
        self.__val = val 
    def get_val(self):
        return self.__val 
    
son = Child(100, 200) 
print(son.get_num()) 
print(son.get_val())
```

```python
class Parent:
    def __init__(self):
        self.num = 100 

class Child(Parent):
    def __init__(self):
        super().__init__()
        self.var = 200 
    def show(self):
        print(self.num)
        print(self.var) 

son = Child() 
son.show()
```

```python
# single Inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print('Inside phone constructor')
        self.__price = price 
        self.brand = brand 
        self.camera = camera

    def buy(self):
        print('Buying a phone')

class SmartPhone(Phone):
    pass 

SmartPhone(1000, 'apple', '13px').buy()
```

```python
# multilevel 

class Product:
    
    def review(self):
        print('Product customer reviews')

class Phone(Product):
    def __init__(self, price, brand, camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand 
        self.camera = camera 

    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):
    pass 
s = SmartPhone(20000,'apple', 12)
s.review()
s.buy()
```

```python
# Hierarchical

class Phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand 
        self.camera = camera 

    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):
    pass 

class FeaturePhone(Phone):
    pass
s = SmartPhone(20000,'apple', 12).buy()
f = FeaturePhone(20000, 'lava', '23').buy()

```

```python
# Multiple 
class Phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor") 
        self.__price = price 
        self.brand = brand 
        self.camera = camera 

    def buy(self):
        print("Buying a phone")

class Product:
    def review(self):
        print('Customer review')

class SmartPhone(Phone, Product):
    pass

s = SmartPhone(20000, 'apple', 12)
s.buy()
s.review()
```

```python
# The diamond problem
class Phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor") 
        self.__price = price 
        self.brand = brand 
        self.camera = camera 

    def buy(self):
        print("Buying a phone by Phone Class")

class Product:
    def buy(self):
        print('Product buy method')

# Method resolution order
class SmartPhone(Phone, Product):
    pass

s = SmartPhone(20000, 'apple', 12)
s.buy()

```

In multiple inheritance, Python follows MRO.
Only ONE constructor runs unless you explicitly chain them using super().

SO how do we trigger all constructors?

we must use super() properly.

```python
class Phone:
    def __init__(self):
        print("Phone constructor")
        super().__init__()

class Product:
    def __init__(self):
        print("Product constructor")
        super().__init__()

class SmartPhone(Phone, Product):
    def __init__(self):
        print("SmartPhone constructor")
        super().__init__()

s = SmartPhone()
```

```python
class A:
    def m1(self):
        return 20 
class B:
    def m1(self):
        return 30 
    def m2(self):
        return 40
class C(B):
    def m2(self):
        return 20 
obj1 = A()
obj2 = B()
obj3 = C()
print(obj1.m1() + obj2.m1() + obj3.m2())
```

```python
class A:
    def m1(self):
        return 20 

class B(A):
    def m1(self):
        val = super().m1() + 30

class C(B):
    def m1(self):
        val = self.m1() + 20 
        return val 

obj = C()
print(obj.m1())
```

Method overloading is not supported in python

```python
class Shape:
    def area(self, radius):
        print('area of circle')
        return 3.14 * radius * radius

    def area(self, l, b):
        print('area of rectangle')
        return l * b 

area = Shape()
area.area(2)

```

```python
area.area(2, 2)
```

Another way to do method overloading using default parameters

```python
class Shape:
    def area(self, a, b=0):
        if b == 0:
            return 3.14 * a**2 
        else:
            return a * b 
        


area = Shape()
area.area(2)

```

Abstraction 

```python
from abc import ABC, abstractmethod  # Abstract base class

class BankApp(ABC):

    def database(self):
        print('connected to database')

    @abstractmethod # atleast 1 abstract method should be there
    def security(self):
        pass 
```

```python
class MobileApp(BankApp):

    def mobile_login(self):
        print('login into mobile')

mob = MobileApp()

```

```python
class MobileApp(BankApp):

    def mobile_login(self):
        print('login into mobile')

    # hiding details but enforce child class to build this method.
    def security(self):
        print('mobile security')

mob = MobileApp()

```

```python
mob.database()
```
