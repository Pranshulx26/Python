3. Variables

```python
# Static vs Dynamic Typing
# Static vs Dynamic Binding
# Stylish declaration techniques
```

```python
# C/C++
# int name = 'pranshul'
name = 'pranshul'
print(name)

a = 5 
b = 6 
print(a + b)
```

```python
# Dynamic Typing
# int a = 5; No need to declare the type of variable before hand 
# that is called dynamic typing
x = 2.5
print(type(x))
x = 2
print(type(x))
x = 'pranshul'
print(type(x))
print('Automatically infering the data type based on value')
```

```python
# Dynamic Binding
# We can change the data type of any variable at any time that is dynamic binding
a = 5 
print(a)
a = 'pranshul'
print(a)
```

```python
a = 1 
b = 2 
c = 3 
print(a, b, c)
a, b, c = 1, 2, 3
print(a, b, c)
```

```python
a=b=c = 1
print(a, b, c)
```

6. Type Conversion

```python
# Implicit vs Explicit

# Implicit type conversion
print(5 + 5.6)
print(type(5), type(5.6))


# print(4 + '4') -> TypeError

# Explicit Type Conversion
# str -> int 
print(int('23'))

# int -> str
print(str(34))

# flaot -> int 
print(int(234.32))

# int -> float
print(float(23))
```

7. Literals

```python
# any raw value we store in variable is a literal
a = 0b1010 # Binary Literal
b = 100 # Decimal Literal
c = 0o310 # Octal Literal
d = 0x12c # Hexadecimal Literal

# Float Literal
float_1 = 10.5 
float_2 = 1.5e2 
float_3 = 1.5e-3

# Complex Literal
x = 3.14j

print(a, b, c, d)
print(float_1, float_2, float_3)
print(x, x.imag, x.real)
```
