Operators in Python

- Arithmetic Operators
- Relational Operators
- Logical Operators
- Bitwise Operators
- Assignment Operators
- Membership Operators

```python
# Arithmetic operators

print(5 + 6)
print(5 - 6)
print(5 * 6)
print(5 / 6)
print(5 // 2)
print(5 % 2)
print(5 ** 2)
```

```python
# Relational Operators
print(4 > 5)
print(4 < 5)
print(4 >= 5)
print(4 <= 5)
print(5 == 5)
print(5 != 5)
```

```python
# Logical Operators
print(1 and 0)

print(1 or 0)

print(not 1)

```

```python
# Bitwise Operators

# bitwise and 
print(2 & 3)
print(0b10 & 0b11)
# 1 0
# 1 1
# ---
# 1 0 = 2

# bitwise or 
print(2 | 3)
print(0b10 | 0b11)
# 1 0
# 1 1
# ---
# 1 1 = 3

# bitwise xor 
print(2 ^ 3)
# 1 0
# 1 1
# ---
# 0 1 = 1 # dono same -> 0 different -> 1

# bitwise not (~)
print(~3)

# left shift
print(4 >> 2) 

# right shift
print(5 << 2)




```

```python
# Assignment operator
# = 
# a = 2 

a = 2 
a += 2  # a = a + 2 
print(a)
```

```python
# Membership operator

# in / not in 

print('D' in 'Delhi')
print('D' not in 'Delhi')
```

```python
# program - Find the sum of a 3 digit number entered by the user
number = input("Enter 3 digit number: ")

res = 0 
for num in number:
    res += int(num)
print(res)


res = 0 
number = int(number)
while number:
    res += number % 10 
    number = number // 10
print(res)

```

if-else in python

```python
# Login program and indentation



# if condition:
#     # code 
# else:
#     # code 

email = input('Enter Email: ')
password = input('Enter Password')

if email == 'pranshul@gmail.com' and password == 'xyz':
    print('Login Successful') 
elif email == 'pranshul@gmail.com' and password != 'xyz':
    print('Incorrect password')
    password = input("Enter password again")
    if password == 'xyz':
        print('Login Successful')
    else:
        print('Wrong password')
elif email != 'pranshul@gmail.com' and password == 'xyz':
    print('Incorrect email')
    email = input("Enter email again")
    if email == 'pranshul@gmail.com':
        print('Login Successful')
    else:
        print('Wrong email')
else:
    print('Not correct')
    

```

```python
# if-else examples
# 1. Find the min of 3 given numbers

a = int(input('First number'))
b = int(input('Second number'))
c = int(input('Third number'))

if a < b and a < c :
    print("smallest is ", a)
elif b < c:
    print("smallest is ", b)
else:
    print('smallest is ', c)
```

```python
# 2. menu driven calculator
fnum = int(input("Enter 1st number"))
snum = int(input("Enter 2nt number"))

op = input("Enter the operation")

if op == '+':
    print(fnum + snum)
elif op == '-':
    print(fnum - snum)
else:
    print(fnum * snum)
```

```python
# while loop with else
x = 1 

while x < 3:
    print(x)
    x += 1
else:
    print('Limit crossed')
```

```python
# guessing game

chances = 5 
random_number = random.randint(1, 100)
print("You have only 5 chances to win this game.")
while chances > 0:
    guess = int(input("Guess number between 1 and 100"))
    if guess == random_number:
        print('Hurray!! Congratulations you won!')
        print(f'You take {5 - chances} chances to win this game')
        break 
    elif guess < random_number:
        print('Guess Higher')
        chances -= 1
        print(f'You have {chances} chance left ')
        if chances == 0:
            print('You loose')
    else:
        print('Guess Lower')
        chances -= 1
        print(f'You have {chances} chance left ')
        if chances == 0:
            print('You loose')



```
