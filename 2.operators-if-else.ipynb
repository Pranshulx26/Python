Operators in Python

- Arithmetic Operators
- Relational Operators
- Logical Operators
- Bitwise Operators
- Assignment Operators
- Membership Operators

```python
# Arithmetic operators

print(5 + 6)
print(5 - 6)
print(5 * 6)
print(5 / 6)
print(5 // 2)
print(5 % 2)
print(5 ** 2)
```

```python
# Relational Operators
print(4 > 5)
print(4 < 5)
print(4 >= 5)
print(4 <= 5)
print(5 == 5)
print(5 != 5)
```

```python
# Logical Operators
print(1 and 0)

print(1 or 0)

print(not 1)

```

```python
# Bitwise Operators

# bitwise and 
print(2 & 3)
print(0b10 & 0b11)
# 1 0
# 1 1
# ---
# 1 0 = 2

# bitwise or 
print(2 | 3)
print(0b10 | 0b11)
# 1 0
# 1 1
# ---
# 1 1 = 3

# bitwise xor 
print(2 ^ 3)
# 1 0
# 1 1
# ---
# 0 1 = 1 # dono same -> 0 different -> 1

# bitwise not (~)
print(~3)

# left shift
print(4 >> 2) 

# right shift
print(5 << 2)




```

```python
# Assignment operator
# = 
# a = 2 

a = 2 
a += 2  # a = a + 2 
print(a)
```

```python
# Membership operator

# in / not in 

print('D' in 'Delhi')
print('D' not in 'Delhi')
```

```python
# program - Find the sum of a 3 digit number entered by the user
number = input("Enter 3 digit number: ")

res = 0 
for num in number:
    res += int(num)
print(res)


res = 0 
number = int(number)
while number:
    res += number % 10 
    number = number // 10
print(res)

```
