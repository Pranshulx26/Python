```python
import pandas as pd 
import numpy as np 
```

```python
# What are vectorized string operations?
# - These are string operations that can be applied to entire Series or DataFrame columns without the need for explicit loops. They are optimized for performance and can handle missing values gracefully.
a = np.array([1, 2, 3, 4])
a * 4
```

```python
# problem in vectorized operations in vanila python 
s = ['cat', 'mat', None]
[i.startswith('c') for i in s]
```

```python
# How pandas solves this issue?

s = pd.Series(['cat', 'mat', None])
# str = string accessor
s.str.startswith('c')

# fast and optimized for larger datasets 
# why is it fast?
# - pandas uses Cython and other optimizations under the hood to speed up string operations. It also leverages vectorized operations, which are implemented in C and can process data in bulk, reducing the overhead of Python loops.
```

```python
# import titanic
df = pd.read_csv('titanic.csv')
df.head()
```

```python
df['Name']
```

```python
# Common functions 
# lower/upper/capitalize/title
df['Name'].str.lower().head()
```

```python
df['Name'].str.upper().head()
```

```python
df['Name'].str.capitalize().head()
```

```python
df['Name'].str.title().head()
```

```python
# len 
df['Name'].str.len().max()
```

```python
df['Name'][df['Name'].str.len() == 82]
```

```python
# strip
string = '   Hello, World!   '.strip()
string
```

```python
# split -> get 
df['lastname'] = df['Name'].str.split(',').str.get(0)
```

```python
df[['title', 'firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n=1, expand=True)
```

```python
df.head()
```

```python
df['title'].value_counts()
```

```python
# replace 
df['title'].str.replace('Ms.', 'Miss.').replace('Mlle.', 'Miss.').value_counts()
```

```python
# filtering 
# startswith/endswith
df[df['firstname'].str.startswith('A')]['firstname']
# isdigit/isalpha/isalnum
df[df['firstname'].str.isdigit()]['firstname']
```

```python
# applying regex 
# contains
# search john -> both case
df[df['firstname'].str.contains('john', case=False)]['firstname'].head()
```

```python
df[df['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$', regex=True)]['lastname'].head()
```
