Series is 1D and DataFrames are 2D objects 


```python
# can we have multiple index? let's try  
index_val = [('cse', 2019), ('cse', 2020), ('cse', 2021), ('cse', 2022), ('ece', 2019), ('ece', 2020), ('ece', 2021), ('ece', 2022)]
```

```python
a = pd.Series(data=np.random.randint(1, 100, 8), index=index_val)
a
```

```python
a[('cse', 2019)]
```

```python
# The problem? 
# a['cse'] # Error
```

```python
# The solution? MultiIndex objects!
# multiple index levels within a single index object.
```

```python
# How to create mutliindex objects?
# 1. pd.MultiIndex.from_tuples()
index_val = [('cse', 2019), ('cse', 2020), ('cse', 2021), ('cse', 2022), ('ece', 2019), ('ece', 2020), ('ece', 2021), ('ece', 2022)]
multiindex = pd.MultiIndex.from_tuples(index_val, names=['dept', 'year'])
multiindex
```

```python
s = pd.Series(data=np.random.randint(1, 100, 8), index=multiindex)
s
```

```python
s['cse']
```

```python
# 2. pd.MultiIndex.from_product()

pd.Series(np.random.randint(1, 100, 8), index=pd.MultiIndex.from_product([['cse', 'ece'], [2019, 2020, 2021, 2022]], names=['dept', 'year']))
```

```python
# create series with multiindex 
s = pd.Series([1, 2, 3, 4, 5, 6, 7, 8], index=multiindex)
```

```python
s
```

```python
# how to fetch items from such a series?
s[('cse', 2019)]
```

```python
s['cse']
```

```python
# unstack
s.unstack()
```

```python
s.unstack(level=0)
```

```python
# stack 
s.unstack().stack()
```

Long vs Wide Data

Wide format is where we have a single row for every data point with multiple columns to hold the valeus of various attributes.

Long format is where, for each data point we have as many rows as the number of attributesand each row contains the value of a particular attribute for a given data point.

```python
# melt -> wide to long format 
pd.DataFrame({'cse': [120]})
```

```python
pd.DataFrame({'cse': [120]}).melt()
```

```python
pd.DataFrame({'cse': [120], 'ece': [100], 'mech': [50]})
```

```python
pd.DataFrame({'cse': [120], 'ece': [100], 'mech': [50]}).melt(var_name='branch', value_name='students')
```

```python
pd.DataFrame(
    {
        'branch': ['cse', 'ece', 'mech'],
        '2020': [100, 150, 60],
        '2021': [120, 130, 80],
        '2022': [130, 120, 90]
    }
)
```

```python
pd.DataFrame(
    {
        'branch': ['cse', 'ece', 'mech'],
        '2020': [100, 150, 60],
        '2021': [120, 130, 80],
        '2022': [130, 120, 90]
    }
).melt(id_vars=['branch'], var_name='year', value_name='students')
```

```python
death = pd.read_csv('time_series_covid19_deaths_global.csv')
confirm = pd.read_csv('time_series_covid19_confirmed_global.csv')
```

```python
death.head()
```

```python
death.shape
```

```python
confirm.head()
```

```python
death = death.melt(id_vars=['Province/State', 'Country/Region', 'Lat', 'Long'], var_name='date', value_name='death_count')
confirm = confirm.melt(id_vars=['Province/State', 'Country/Region', 'Lat', 'Long'], var_name='date', value_name='confirm_count')
```

```python
confirm.merge(death, on=['Province/State', 'Country/Region', 'Lat', 'Long', 'date'], how='inner')
```
