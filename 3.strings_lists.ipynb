```python
n = int(input("Enter a number"))

result = 0 
fact = 1   
for i in range(1, n+1):
    fact = fact * i 
    result = result +  i / fact
print(result)

```

```python
rows = int(input("Enter number of rows: "))

for i in range(1, rows+1):
    for j in range(1, i+1):
        print(j, end='')
    for k in range(i-1, 0, -1):
        print(k, end='')
    print()

```

```python
# print prime numbers between given range 

lower = int(input("Enter lower range: "))
upper = int(input("Enter upper range: "))

for i in range(lower, upper+1):
    for j in range(2, i):
        if i % j == 0:
            break 
    else:
        print(i)
```

Example Programs

```python
# Find the length of a given string without using the len() function
string = input('Enter string to find length: ')
counter = 0 
for i in string:
    counter += 1 
print("Length of ", string, 'is ', counter)
```

```python
# Extract username from a given email.
# Eg if the email is pranshul24@gmail.com
# then the username should be pranshul24

email = input("Enter email to find the username: ")
if '@' not in email:
    print('Enter correct email')
else:
    print('username: ', email.split('@')[0])
```

```python
# Count the frequency of a particular character in a provided string.
# Eg. 'hello how are you' is the string, the frequency of h in this string is 2.

string = input('Enter string')
char_to_find = input("Enter character to find the frequency: ")
print(string.count(char_to_find))
```

```python
# Write a program which can remove a particular character from a string 

string = input('Enter string')
char_to_remove = input("Enter character to remove: ")
new_string = ""
for i in string:
    if i == char_to_remove:
        continue 
    else:
        new_string += i
print(new_string)
```

```python
# Write a program that can check whether a given string is palindrome or not.

string = input('Enter string here: ')
if string == string[::-1]:
    print('It is palindrome')
else:
    print('It is not a palindrome')
```

```python
# Write a program to count the number of words in a string without split()

string = input("Enter string here: ")
counter = 1
for i in string:
    if i == " ":
        counter+=1
print(counter)
```

```python
# Write a python program to convert a string to title case without using a title()

string = input("Enter string: ")
dummy = ''
lst = []
string_split = string.split()
for word in string_split:
    for i in range(len(word)):
        if i == 0:
            dummy += word[i].upper()
        else:
            dummy += word[i]
    lst.append(dummy)
    dummy = ''
print(' '.join(lst))
```

```python
# Write a program that can convert an integer to string.

integer = int(input("Enter number here: "))

s = ''
digits = '0123456789'
while integer != 0:
    s = digits[integer % 10] + s # 1234
    integer = integer // 10 
print(s)
```
